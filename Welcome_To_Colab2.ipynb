```

# ============================================================
# DATA SCIENCE MINI PROJECT
# STUDENT PERFORMANCE ANALYSIS
# ============================================================

# 1. IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import io

from google.colab import files

print("All Libraries Imported Successfully!")


# ============================================================
# 2. UPLOAD DATASET
# ============================================================

uploaded = files.upload()

filename = next(iter(uploaded))

df = pd.read_csv(io.BytesIO(uploaded[filename]))

print("\nDataset Uploaded Successfully!")

print(df.head())


# ============================================================
# 3. DATA EXPLORATION
# ============================================================

print("\n========== DATA EXPLORATION ==========")

print("First 5 Rows:")
print(df.head())

print("\nLast 5 Rows:")
print(df.tail())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe(include="all"))


# ============================================================
# 4. DATA QUALITY CHECK
# ============================================================

print("\n========== DATA QUALITY ==========")

print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nUnique Values:")
print(df.nunique())


# ============================================================
# 5. DATA CLEANING
# ============================================================

# Remove duplicate records

df = df.drop_duplicates().copy()

# Handle missing numerical values

numeric_cols = df.select_dtypes(
    include=np.number
).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Handle missing categorical values

categorical_cols = df.select_dtypes(
    include="object"
).columns

for col in categorical_cols:
    mode_values = df[col].mode()

    if not mode_values.empty:
        df[col] = df[col].fillna(mode_values.iloc[0])

print("\nData Cleaning Completed!")

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicates After Cleaning:")
print(df.duplicated().sum())


# ============================================================
# 6. FEATURE ENGINEERING
# ============================================================

score_cols = [
    "math score",
    "reading score",
    "writing score"
]

df["Average Score"] = df[score_cols].mean(axis=1)

df["Total Score"] = df[score_cols].sum(axis=1)

df["Performance"] = np.where(
    df["Average Score"] >= 70,
    "Good",
    "Needs Improvement"
)

print("\nNew Features Created:")
print(df.head())


# ============================================================
# 7. BASIC DATA ANALYSIS
# ============================================================

print("\n========== BASIC ANALYSIS ==========")

print("Total Students:", len(df))

print("Average Math Score:",
      round(df["math score"].mean(), 2))

print("Average Reading Score:",
      round(df["reading score"].mean(), 2))

print("Average Writing Score:",
      round(df["writing score"].mean(), 2))

print("Overall Average Score:",
      round(df["Average Score"].mean(), 2))

print("Highest Score:",
      df[score_cols].max().max())

print("Lowest Score:",
      df[score_cols].min().min())


# ============================================================
# 8. GENDER-WISE ANALYSIS
# ============================================================

print("\n========== GENDER ANALYSIS ==========")

gender_analysis = df.groupby("gender")[
    score_cols
].mean()

print(gender_analysis)


# ============================================================
# 9. TEST PREPARATION ANALYSIS
# ============================================================

print("\n========== TEST PREPARATION ==========")

preparation_analysis = df.groupby(
    "test preparation course"
)[score_cols].mean()

print(preparation_analysis)


# ============================================================
# 10. PARENTAL EDUCATION ANALYSIS
# ============================================================

print("\n========== PARENTAL EDUCATION ==========")

parent_analysis = df.groupby(
    "parental level of education"
)["Average Score"].mean().sort_values(
    ascending=False
)

print(parent_analysis)


# ============================================================
# 11. TOP 10 STUDENTS
# ============================================================

print("\n========== TOP 10 STUDENTS ==========")

top_students = df.nlargest(
    10, "Average Score"
)

print(top_students[
    ["gender", "math score",
     "reading score", "writing score",
     "Average Score"]
])


# ============================================================
# 12. BAR CHART
# ============================================================

plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x="gender",
    y="Average Score"
)

plt.title("Average Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")

plt.show()


# ============================================================
# 13. LINE CHART
# ============================================================

subject_average = df[score_cols].mean()

plt.figure(figsize=(8,5))

plt.plot(
    ["Math", "Reading", "Writing"],
    subject_average.values,
    marker="o"
)

plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")

plt.show()


# ============================================================
# 14. PIE CHART
# ============================================================

performance_count = df["Performance"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    performance_count,
    labels=performance_count.index,
    autopct="%1.1f%%"
)

plt.title("Student Performance Distribution")

plt.show()


# ============================================================
# 15. HISTOGRAM
# ============================================================

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Average Score",
    bins=10
)

plt.title("Average Score Distribution")
plt.xlabel("Average Score")
plt.ylabel("Student Count")

plt.show()


# ============================================================
# 16. SCATTER PLOT
# ============================================================

plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="math score",
    y="reading score",
    hue="gender"
)

plt.title("Math vs Reading Score")

plt.show()


# ============================================================
# 17. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(8,6))

sns.heatmap(
    df[score_cols + ["Average Score"]].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Student Score Correlation")

plt.show()


# ============================================================
# 18. KEY INSIGHTS
# ============================================================

print("\n========== KEY INSIGHTS ==========")

print("Overall Average:",
      round(df["Average Score"].mean(), 2))

print("Best Subject:",
      df[score_cols].mean().idxmax())

print("Lowest Average Subject:",
      df[score_cols].mean().idxmin())

print("\nPerformance Distribution:")
print(df["Performance"].value_counts())

print("\nGender-wise Average:")
print(gender_analysis)

print("\nTest Preparation Results:")
print(preparation_analysis)

print("\nParental Education Results:")
print(parent_analysis)


# ============================================================
# 19. SAVE CLEAN DATASET
# ============================================================

df.to_csv(
    "clean_student_performance.csv",
    index=False
)

print("\nClean Dataset Saved Successfully!")


# ============================================================
# 20. DOWNLOAD CLEAN DATASET
# ============================================================

files.download("clean_student_performance.csv")


print("\n===================================")
print("DATA SCIENCE MINI PROJECT COMPLETED")
print("===================================")
```
